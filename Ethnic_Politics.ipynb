Libraries and Input

```python
import pandas as pd
import plotly.express as px

# CSV file path to EPR data (stored on my GitHub)
url = "https://raw.githubusercontent.com/Djimdou/Side-Projects/master/EthnicPolitics/EPR-2021.csv"
```

Data Importation

```python
# Read the CSV file from GitHub
df = pd.read_csv(url)

# Show first observations
df.head()
```

```python
# # Chad Data

# Filter on Chad data
df_chad = df[df['statename'] == 'Chad']

 # For later graph, need to convert types of the dates
df_chad['from'] = pd.to_datetime(df_chad['from'], format='%Y', errors='coerce')
df_chad['to'] = pd.to_datetime(df_chad['to'], format='%Y', errors='coerce')


# Show firts lines
df_chad.head()
```

Gantt Cahrt

```python
# Create a timeline (Gantt) chart

fig = px.timeline(
    df_chad,
    x_start="from",
    x_end="to",
    y="group",
    color="status",
    title="Evolution of the political status of ethnic groups in Chad",
    color_discrete_map={
        "MONOPOLY": "darkred",
        "DOMINANT": "orange",
        "SENIOR PARTNER": "gold",
        "JUNIOR PARTNER": "lightgreen",
        "POWERLESS": "lightgray",
        "DISCRIMINATED": "black",
        "SELF-EXCLUSION": "blue",
        "REGIONAL AUTONOMY": "blue"
    }
)
fig.update_yaxes(autorange="reversed")  # To make the top of the chart the top of the table
fig.show()
```
